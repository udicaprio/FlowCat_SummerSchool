## 1. Basic Python

A variable gives a name to a value. Python infers the value's type: text is a `str`, a whole number is an `int`, a decimal is a `float`, and `True` or `False` is a `bool`.

```python
reactor_name = "Intensified reactor"
temperature_C = 60.0
residence_time_min = 10
target_reached = True
residence_times_min = [3.2, 4, 6]

print(reactor_name)
print(residence_times_min)
print(temperature_C, type(temperature_C))
print(residence_time_min, type(residence_time_min))
print(target_reached, type(target_reached))
```

### Step 1 — Make a calculation

Convert the temperature from degrees Celsius to kelvin using

$$T_K = T_{°C} + 273.15$$

```python
temperature_K = temperature_C + 273.15
print("Operating temperature:", temperature_K, "K")
```

### Step 2 — Use a conditional statement

An `if` statement runs code only when a condition is true. Add an `else` branch so the code also reports when the conversion target is not reached. Indentation is part of Python's syntax.

```python
conversion_percent = 82.6
target_conversion_percent = 80.0

if conversion_percent >= target_conversion_percent:
    print("The conversion target is reached.")
else:
    print("The conversion target is not reached.")
```

### Step 3 — Store several values and repeat an action

A list stores an ordered collection. A `for` loop visits every item in that collection. Print each residence time and its corresponding space velocity, $1/\tau$.

```python
for residence_time in residence_times_min:
    space_velocity_per_min = 1 / residence_time
    print("Residence time:", residence_time, "min; space velocity:", space_velocity_per_min, "min^-1")
```

### Step 4 — Define and call functions

A function groups instructions under a reusable name. The first example adds two numbers. The second function calculates and returns the mean, minimum, and maximum of a list.

```python
def add_numbers(first_number, second_number):
    result = first_number + second_number
    return result

addition_result = add_numbers(4, 6)
print("Addition result:", addition_result)

def calculate_statistics(values):
    mean_value = sum(values) / len(values)
    minimum_value = min(values)
    maximum_value = max(values)
    return mean_value, minimum_value, maximum_value

mean_residence_time, minimum_residence_time, maximum_residence_time = calculate_statistics(residence_times_min)

print("Mean residence time:", round(mean_residence_time, 2))
print("Minimum residence time:", minimum_residence_time)
print("Maximum residence time:", maximum_residence_time)
```

## 2. Numerical calculations with NumPy

NumPy provides fast operations on arrays. Unlike a Python list, a NumPy array can perform the same calculation on all its values at once. The conventional and intensified reactor were evaluated at five temperatures.

```python
import numpy as np

temperatures_C = np.array([40, 50, 60, 70, 80])
conversion_conventional = np.array([54.1, 62.4, 70.8, 77.5, 82.1])
conversion_intensified = np.array([66.8, 75.2, 82.6, 88.1, 91.7])

print(temperatures_C)
print(temperatures_C.shape)
```

### Step 5 — Perform element-wise operations

Calculate the conversion improvement at each temperature and convert percentages to fractions.

```python
conversion_improvement = conversion_intensified - conversion_conventional
conversion_fraction = conversion_intensified / 100

print("Improvement in percentage points:", conversion_improvement)
print("Conversion as a fraction:", conversion_fraction)
```

### Step 6 — Summarize an array

Use NumPy to calculate the mean, minimum, maximum, and standard deviation of the intensified-reactor conversion.

```python
mean_conversion = np.mean(conversion_intensified)
minimum_conversion = np.min(conversion_intensified)
maximum_conversion = np.max(conversion_intensified)
std_conversion = np.std(conversion_intensified)

print("Mean conversion [%]:", mean_conversion)
print("Minimum conversion [%]:", minimum_conversion)
print("Maximum conversion [%]:", maximum_conversion)
print("Standard deviation [percentage points]:", std_conversion)
```

### Step 7 — Work with a two-dimensional array

Here, rows represent reactor types and columns represent temperatures. `axis=0` reduces the rows and returns one result per **column**. `axis=1` reduces the columns and returns one result per **row**.

```python
conversion_matrix = np.array([
    conversion_conventional,
    conversion_intensified,
])

mean_at_each_temperature = np.mean(conversion_matrix, axis=0)
mean_for_each_reactor = np.mean(conversion_matrix, axis=1)

print("Matrix shape:", conversion_matrix.shape)
print("Mean at each temperature:", mean_at_each_temperature)
print("Mean for each reactor:", mean_for_each_reactor)
```

`np.linspace(start, stop, number)` is useful for defining evenly spaced operating conditions.

```python
temperature_grid_C = np.linspace(40, 80, 9)
print(temperature_grid_C)
```

## 4. Data visualization with Matplotlib

Plots make patterns and unusual observations easier to see. Every engineering plot should identify the variables and units on both axes.

```python
from matplotlib import pyplot as plt
```

### Step 11 — Create a scatter plot

Plot conversion against residence time, then add axis labels, a title, and a grid.

```python
plt.scatter(df["residence_time_min"], df["conversion_percent"], alpha=0.75)
plt.xlabel("Residence time [min]")
plt.ylabel("Conversion [%]")
plt.title("Conversion versus residence time")
plt.grid(alpha=0.25)
plt.show()
```

### Step 12 — Compare reactor types

Plot each reactor type separately and use a legend. This prevents the two groups from being hidden in one set of points.

```python
for reactor_type, group in df.groupby("reactor_type"):
    plt.scatter(
        group["residence_time_min"],
        group["conversion_percent"],
        label=reactor_type,
        alpha=0.8,
    )

plt.xlabel("Residence time [min]")
plt.ylabel("Conversion [%]")
plt.title("Reactor comparison")
plt.legend()
plt.grid(alpha=0.25)
plt.show()
```

### Step 13 — Plot a distribution

Use a histogram to see how conversion values are distributed.

```python
plt.hist(df["conversion_percent"], bins=8, edgecolor="black")
plt.xlabel("Conversion [%]")
plt.ylabel("Number of runs")
plt.title("Distribution of conversion")
plt.show()
```
